```python
from huggingface_hub import HfApi
from huggingface_hub import get_paths_info
import json

api = HfApi()

repo_id = "allenai/c4"
repo_type = "dataset"
chunk = ["README.md", "en"]

path_infos = list(api.get_paths_info(
    repo_id=repo_id,
    repo_type=repo_type,
    paths=chunk,
    expand=True,     # important: adds size, LFS info, etc.
    token=None       # or your HF token if needed
))

for p in path_infos:
    print(json.dumps(vars(p), indent=2))
```
