```python
from huggingface_hub import list_datasets

task_tag_count = 0

for d in list_datasets(expand=["tags"]):
    tags = d.tags or []

    for tag in tags:
        if tag.startswith("task_categories:"):
            task_tag_count += 1
            break

print(task_tag_count)
```

```python
from huggingface_hub import list_models

pipeline_count = 0

for m in list_models(expand=["pipeline_tag"], limit=None):
    if m.pipeline_tag:
        pipeline_count += 1

print(pipeline_count)
```
